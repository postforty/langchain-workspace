```python
# 상태
from typing import TypedDict

class State(TypedDict):
    message: str
```

```python
# 노드
def greet(state: State):
    greet_message = f"안녕하세요, {state['message']}님!"
    print("greet_message:", greet_message)
    return {"message": greet_message}
```

```python
from langgraph.graph import StateGraph, START, END

graph_builder = StateGraph(State)

graph_builder.add_node("인사_노드", greet)

# 엣지
graph_builder.add_edge(START, "인사_노드")
graph_builder.add_edge("인사_노드", END)

graph = graph_builder.compile()
```

```python
graph
```

```python
result = graph.invoke({"message": "김일남"})
```
