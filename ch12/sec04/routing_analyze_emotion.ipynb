```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model("google_genai:gemini-3.1-flash-lite")
```

```python
from pydantic import BaseModel, Field

class EmotionBotState(BaseModel):
    user_message: str = Field(default="", description="사용자 입력 메시지")
    emotion: str = Field(default="", description="분석된 감정")
    response: str = Field(default="", description="최종 응답 메시지")
```

```python
from langchain.messages import SystemMessage, HumanMessage

def analyze_emotion(state: EmotionBotState):
    message = state.user_message

    messages = [
        SystemMessage(
            content=("당신은 감정 분석 전문가입니다."
                    "사용자의 메시지를 분석하여 'positive', 'negative', 'neutral' 중 하나로 감정을 평가하세요."
                    "답변은 반드시 하나의 단어만 출력하세요.")
        ),
        HumanMessage(content=f"다음 메시지의 감정을 분석해주세요: {message}")
    ]

    response = llm.invoke(messages)
    emotion = response.content.strip().lower()

    return {"emotion": emotion}
```

```python
def generate_positive_response(state: EmotionBotState):
    response = "😄"
    return {"response": response}
```

```python
def generate_negative_response(state: EmotionBotState):
    response = "😨"
    return {"response": response}
```

```python
def generate_neutral_response(state: EmotionBotState):
    response = "🙂"
    return {"response": response}
```

```python
def route_by_emotion(state: EmotionBotState):
    emotion = state.emotion

    if emotion == "positive":
        return "positive_node"
    elif emotion == "negative":
        return "negative_node"
    elif emotion == "neutral":
        return "neutral_node"
```
