```python
from dotenv import load_dotenv
load_dotenv(override=True)
```

```python
from langchain.agents import create_agent
from langchain.agents.middleware import PIIMiddleware

agent = create_agent(
    model="google_genai:gemini-3-flash-preview",
    middleware=[
        PIIMiddleware(
            pii_type="email",
            strategy="redact",
            apply_to_input=True,
            apply_to_output=True
        )
    ]
)
```

```python
prompt = "안녕하세요. 제 이메일은 kim1@example.com 입니다."
```

```python
res = agent.invoke({"messages": [{"role": "user", "content": prompt}]})
```

```python
from langchain.agents import create_agent
from langchain.agents.middleware import PIIMiddleware

agent = create_agent(
    model="google_genai:gemini-3-flash-preview",
    middleware=[
        # 정규 표현식(regex)
        PIIMiddleware(
            "phone_number",
            detector=r"\+?\d{1,3}[\s.-]?\d{3,4}[\s.-]?\d{4}",
            strategy="mask",
        ),
    ]
)
```

```python
prompt = "안녕하세요. 제 전화번호는 010-1234-5678 입니다."
```

```python
res = agent.invoke({"messages": [{"role": "user", "content": prompt}]})
```
