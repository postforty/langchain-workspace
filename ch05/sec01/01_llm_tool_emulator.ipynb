```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.tools import tool

@tool
def get_weather(location: str) -> str:
    """특정 위치의 현재 날씨 정보를 가져옵니다."""
    return f"{location}의 날씨"
```

```python
from langchain.agents import create_agent
from langchain.agents.middleware import LLMToolEmulator

agent = create_agent(
    model="google_genai:gemini-3.1-flash-lite",
    tools=[get_weather],
    # 실제 도구 실행하지 않고 에이전트 동작 테스트
    # 외부 도구 사용할 수 없거나 비용이 많이 들 때 테스트
    # 실제 도구 구현전 에이전트 워크플로 프로토타입 제작 가능
    middleware=[LLMToolEmulator(model="google_genai:gemini-3.1-flash-lite")]
)
```

```python
response = agent.invoke(
    {"messages": [{"role": "user", "content": "부산 날씨가 어때?"}]},
)

response
```

```python
from langchain.tools import tool

@tool
def send_email(to: str, subject: str, body: str) -> str:
    """이메일을 발송합니다."""
    return "이메일 발송 완료"
```

```python
from langchain.agents import create_agent
from langchain.agents.middleware import LLMToolEmulator

agent = create_agent(
    model="google_genai:gemini-3.1-flash-lite",
    tools=[get_weather, send_email],
    middleware=[LLMToolEmulator(model="google_genai:gemini-3.1-flash-lite")]
)
```

```python
prompt = """김일남님께 이메일 발송해줘.

- 이메일 주소: kim1@example.com
- 제목: 랭체인 공부
- 내용: 미들웨어 부분 공부하세요.
"""
```

```python
response = agent.invoke(
    {"messages": [{"role": "user", "content": prompt}]},
)

response
```
