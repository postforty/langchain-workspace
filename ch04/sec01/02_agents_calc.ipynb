```python
from langchain.tools import tool

@tool
def add(a: int, b: int) -> int:
    """
    두 숫자를 더해 그 합을 반환합니다.
    
    Args:
        a: 더할 첫 번째 숫자
        b: 더할 두 번째 숫자
    """
    return a + b
```

```python
@tool
def subtract(a: int, b: int) -> int:
    """
    첫 번째 숫자에서 두 번째 숫자를 뺀 값을 반환합니다.
    
    Args:
        a: 빼기 연산의 대상이 되는 수 (피감수)
        b: 뺄 숫자 (감수)
    """
    return a - b
```

```python
@tool
def multiply(a: int, b: int) -> int:
    """
    두 숫자를 곱해 그 결과를 반환합니다.
    
    Args:
        a: 곱할 첫 번째 숫자
        b: 곱할 두 번째 숫자
    """
    return a * b
```

```python
@tool
def divide(a: int, b: int) -> float:
    """
    첫 번째 숫자를 두 번째 숫자로 나눕니다.
    
    Args:
        a: 나뉘는 수 (피제수)
        b: 나누는 수 (제수)
    """
    return a / b
```

```python
tools = [add, subtract, multiply, divide]
```

```python
from langchain.agents import create_agent

agent = create_agent(
    "google_genai:gemini-3.1-flash-lite",
    system_prompt="사칙 연산 도구를 사용할 때는 반드시 연산자 우선 순위에 맞게 연산합니다.",
    tools=tools)
```

```python
agent.invoke(
    {"messages": [{"role": "user", "content": "1234 + 4568 * 7890 / 123 = "}]}
)
```
