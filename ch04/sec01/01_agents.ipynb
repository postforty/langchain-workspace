# Agents

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.tools import tool

@tool
def get_weather(city: str) -> str:
    """특정 도시의 날씨를 가져옵니다."""
    return f"{city}: 폭우"
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model="google_genai:gemini-2.5-flash-lite",
    tools=[get_weather],
    system_prompt="당신은 아주 냉철한 기상 캐스터입니다."
)
```

```python
agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "오늘 부산의 날씨를 알려줘"
        }
    ]}
)
```

```python
from langchain.agents import create_agent
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=2.0,
    max_tokens=1000,
    timeout=30
)

agent = create_agent(
    model,
    tools=[get_weather],
    system_prompt="당신은 아주 귀엽고 상큼한 기상 캐스터입니다."
)
```

```python
agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "오늘 부산의 날씨를 알려줘"
        }
    ]}
)
```

### 사칙 연산 툴

```python
from langchain.tools import tool

@tool
def add(a: int | float, b: int | float) -> int | float:
    """
    두 숫자를 더해 그 합을 반환합니다.

    Args:
        a: 더할 첫 번째 숫자
        b: 더할 두 번째 숫자
    """
    return a + b

@tool
def subtract(a: int | float, b: int | float) -> int | float:
    """
    첫 번째 숫자에서 두 번째 숫자를 뺍니다.

    Args:
        a: 빼기 연산의 대상이 되는 수 (피감수)
        b: 뺄 숫자 (감수)
    """
    return a - b

@tool
def multiply(a: int | float, b: int | float) -> int | float:
    """
    두 숫자를 곱해 그 결과를 반환합니다.

    Args:
        a: 곱할 첫 번째 숫자
        b: 곱할 두 번째 숫자
    """
    return a * b

@tool
def divide(a: int | float, b: int | float) -> float | float | str:
    """
    첫 번째 숫자를 두 번째 숫자로 나눕니다.
    0으로 나누려고 하면 에러 메시지를 반환합니다.

    Args:
        a: 나뉘는 수 (피제수)
        b: 나누는 수 (제수)
    """
    if b == 0:
        return "Error: 0으로는 나눌 수 없습니다."
    return a / b

tools = [add, subtract, multiply, divide]
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model="google_genai:gemini-2.5-flash",
    tools=tools,
    system_prompt="""당신은 유능하고 친절한 수학 선생님입니다.
사칙 연산 요청 시 계산 우선 순위별로 단계를 나누고 계획을 세웁니다.
순서대로 Tool을 이용해 계산한 후 최종 답변을 생성합니다."""
)
```

```python
agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "12345 + 456789 * 789 / 12의 답은 뭐야?"
        }
    ]
})
```
