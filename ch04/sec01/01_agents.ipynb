#### 도구 작성 방법
1. 함수 작성
2. 타입 힌팅
3. docstring 작성
4. @tool 적용

```python
from langchain.tools import tool

@tool
def get_weather(city: str) -> str:
    """특정 도시의 날씨를 가져오는 도구입니다."""
    return f"{city}: 맑음"
```

```python
tools = [get_weather]
```

```python
from langchain.agents import create_agent

agent = create_agent("google_genai:gemini-3.1-flash-lite", tools=tools)
```

```python
agent.invoke(
    {"messages": [{"role": "user", "content": "부산의 오늘 날씨는?"}]}
)
```
