```python
from langchain.chat_models import init_chat_model

model = init_chat_model("google_genai:gemini-2.5-flash-lite")
```

```python
conversation = [
    {"role": "system", "content": "당신은 영어를 한글로 번역하는 친철한 조수입니다."},
    {"role": "user", "content": "I love programming."},
    {"role": "assistant", "content": "프로그래밍을 사랑합니다."},
    {"role": "user", "content": "I love building applications."},
]
```

```python
model.invoke(conversation)
```

```python
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage

conversation = [
    SystemMessage("당신은 영어를 한글로 번역하는 친철한 조수입니다."),
    HumanMessage("I love programming."),
    AIMessage("프로그래밍을 사랑합니다."),
    HumanMessage("I love building applications."),
]
```

```python
model.invoke(conversation)
```

```python
for chunk in model.stream("앵무새는 왜 깃털이 화려할까요?"):
    print(chunk.text, end="|", flush=True)
```

```python
response = model.batch([
    "앵무새는 왜 깃털이 화려한가요?",
    "비행기는 어떻게 하늘을 날까요?",
    "양자 컴퓨팅이란 무엇인가요?"
])

response
```

### 도구 호출

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("google_genai:gemini-2.5-flash-lite")
```

```python
from langchain.tools import tool

# 1. 타입 힌팅
# 2. doc string
# 3. @tool

@tool
def get_weather(location: str) -> str:
    """특정 위치의 날씨를 가져옵니다."""
    return f"{location}의 날씨는 맑습니다."
```

```python
model_with_tools = model.bind_tools([get_weather])
```

```python
model_with_tools.invoke("부산의 날씨는 어떤가요?")
```
