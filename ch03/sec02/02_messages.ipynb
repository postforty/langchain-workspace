```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("google_genai:gemini-2.5-flash-lite")
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("당신은 최고의 개그맨입니다."),
    HumanMessage("랭체인에대한 삼행시를 작성해줘."),
    AIMessage("랭킹 꼴찌...")
]

res = model.invoke(messages)
```

```python
print(res.content)
```

```python
messages = [
    {"role": "system" ,"content": "당신은 최고의 개그맨입니다."},
    {"role": "user" ,"content": "랭체인에대한 삼행시를 작성해줘."},
    {"role": "assistant" ,"content": "랭킹 꼴찌..."}
]

res = model.invoke(messages)
```

```python
print(res.content)
```

```python
from langchain_core.prompts import MessagesPlaceholder

messages = [
    SystemMessage(content="당신은 사용자에게 친절하게 답변하는 에이전트입니다. 이전 대화를 기억하고 답변하세요."),
    MessagesPlaceholder(variable_name="history"),
    HumanMessage(content="{input}")
]
```

```python
sample_history = [
    HumanMessage(content="내 이름은 김일남이야."),
    AIMessage(content="김일남님 반갑습니다."),
]
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(messages)
```

```python
chain = prompt | model
```

```python
chain.invoke({
    "history": sample_history,
    "input": "내 이름은 뭐야?"
})
```
