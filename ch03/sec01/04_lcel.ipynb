```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.prompts import PromptTemplate

template = "{language} 할 수 있어?"

prompt = PromptTemplate.from_template(template)
```

```python
prompt.invoke("한국어")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite-preview",
)
```

```python
model.invoke("한국어 할 수 있어?")
```

```python
# LCEL(LangChain Expression Language)
chain = prompt | model # 워크 플로우 구성
```

```python
chain
```

```python
chain.invoke("한국어")
```

```python
chain.invoke("일본어")
```

```python
template = "{topic}에 대해 아주 쉽게 설명해줘"
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(template)
```

```python
prompt
```

```python
prompt.invoke("LangChain")
```

```python
prompt.stream("LangChain")
```

```python
chain = prompt | model
```

```python
user_input = {"topic": "LangChain"}
```

```python
response = chain.stream(user_input)
```

```python
response
```

```python
next(response)
```

```python
for token in response:
    print(token.content[0].get("text"), end="", flush=True)
```

```python
from langchain_core.output_parsers import StrOutputParser

output_parser = StrOutputParser()
```

```python
chain = prompt | model | output_parser
```

```python
response = chain.stream(user_input)
```

```python
next(response)
```

```python
for token in response:
    print(token, end="", flush=True)
```
