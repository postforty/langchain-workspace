```python
from langchain_core.messages import SystemMessage
from langchain_core.prompts import HumanMessagePromptTemplate, MessagesPlaceholder

template = [
    SystemMessage(content="너는 유치원생이야. 유치원생 처럼 답변해줘. 앞선 대화 패턴에 맞게 답변할 것"),
    MessagesPlaceholder(variable_name="history"),
    HumanMessagePromptTemplate.from_template("{value}"),
]
```

```python
template
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(template)
```

```python
response = prompt.invoke({
    "value": "오리",
    "history": [
        {"role": "human", "content": "참새"},
        {"role": "assistant", "content": "짹짹"},
        {"role": "human", "content": "고양이"},
        {"role": "assistant", "content": "냐옹"},
        {"role": "human", "content": "강아지"},
        {"role": "assistant", "content": "멍멍"},
    ]
})
```

```python
response
```
