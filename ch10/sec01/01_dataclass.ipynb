```python
class Context:
    def __init__(self, user_name: str, age: int):
        self.user_name = user_name
        self.age = age
    
    def __eq__(self, other):
        if not isinstance(other, Context):
            return NotImplemented
        return (self.user_name == other.user_name and self.age == other.age)
    
    def __repr__(self):
        return f"Context(user_name='{self.user_name}', age={self.age})"
```

```python
Context("김일남", 99) == Context("김일남", 99)
```

```python
Context("김일남", 99)
```

```python
from dataclasses import dataclass

@dataclass
class Context:
    user_name: str
    age: int
```

```python
Context("김일남", 99) == Context("김일남", 99)
```

```python
Context("김일남", 99)
```
