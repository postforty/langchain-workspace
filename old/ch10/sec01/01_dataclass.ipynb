```python
class Context:
    def __init__(self, user_name: str, age: int = 99):
        self.user_name = user_name
        self.age = age
    
    def __eq__(self, other):
        if not isinstance(other, Context):
            return NotImplemented
        return (self.user_name == other.user_name and self.age == other.age)
    
    def __repr__(self):
        return f"Context(user_name='{self.user_name}', age={self.age})"
```

```python
ctx1 = Context("김일남", 99)
ctx1
```

```python
ctx2 = Context("김일남", 99)
ctx2
```

```python
ctx1 == ctx2
```

```python
from dataclasses import dataclass

@dataclass
class Context:
    user_name: str
    age: int
```

```python
ctx1 = Context("김일남", 99)
ctx1
```

```python
ctx2 = Context("김일남", 99)
ctx2
```

```python
ctx1 == ctx2
```
