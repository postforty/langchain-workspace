```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("google_genai:gemini-2.5-flash")
```

```python
from langchain_core.messages import HumanMessage

model.invoke([HumanMessage('버터떡이 뭐지?')])
```

```python
from langchain.tools import tool
from langchain_community.utilities import DuckDuckGoSearchAPIWrapper
from langchain_community.tools import DuckDuckGoSearchResults

@tool
def get_web_search(query: str, search_period: str='w', region: str='kr-kr') -> str:
    """특정 지역과 기간을 설정하여 웹 검색을 수행합니다.
    
    Args:
        query: 검색어
        search_period: 검색 기간. 'd'(일간), 'w'(주간), 'm'(월간), 'y'(연간) 중 선택
        region: 검색 지역 코드. 한국('kr-kr'), 미국('us-en'), 일본('jp-jp'), 영국('uk-en') 등
    
    Returns:
        문자열(str)로된 검색 결과 리스트
    """
    wrapper = DuckDuckGoSearchAPIWrapper(
        region=region,
        time=search_period
    )

    search = DuckDuckGoSearchResults(
        api_wrapper=wrapper,
        results_separator=";\n"
    )

    return search.invoke(query)
```

```python
from langchain.agents import create_agent

agent = create_agent(model="google_genai:gemini-2.5-flash", tools=[get_web_search])
```

```python
from langchain.messages import HumanMessage

agent.invoke({
    "messages": [HumanMessage(content='버터떡이 뭐지?')]
})
```
