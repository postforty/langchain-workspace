```python
from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(
    model="bge-m3:latest",
)
```

```python
from langchain_postgres import PGVector

# See docker command above to launch a postgres instance with pgvector enabled.
connection = "postgresql+psycopg://langchain:langchain@localhost:5432/vector_db"  # Uses psycopg3!
collection_name = "crud_example_collection"

vector_store = PGVector(
    embeddings=embeddings,
    collection_name=collection_name,
    connection=connection,
    use_jsonb=True,
)
```

##### 1. Create

```python
texts = [
    "강아지는 충성심이 강한 동물입니다.",
    "고양이는 독립적인 성향을 가진 동물입니다.",
    "햄스터는 작고 귀여운 반려동물입니다.",
]

metadatas = [
    {"category": "mammal"},
    {"category": "mammal"},
    {"category": "small_animal"},
]

ids = ["dog-001", "cat-001", "ham-001"]

```

```python
vector_store.add_texts(texts=texts, metadatas=metadatas, ids=ids)
```

---

##### 2. Read

```python
query = "집에서 키우는 작은 동물이 뭐야?"
```

```python
results = vector_store.similarity_search(query, k=1)
```

```python
results
```

---

##### 3. Update

```python
updated_text = ["강아지는 충성심이 아주 강력하며 사람을 잘 따릅니다."]
```

```python
vector_store.add_texts(texts=updated_text, ids=["dog-001"], metadatas=[{"category": "mammal", "updated": True}])
```

---

##### 4. Delete

```python
vector_store.delete(ids=["ham-001"])
```
