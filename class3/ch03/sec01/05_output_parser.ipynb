### 영어 회화 예제

```python
template = """
당신은 영어를 아주 쉽게 가르치는 친절한 영어 선생님입니다.
주어진 상황에 맞는 영어 회화를 작성해 주세요.
양식은 [FORMAT]을 참고해 주세요.

# 상황:
{question}

# Format:
- 영어 회화:
- 한글 해석:
"""
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(template)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash"
)
```

```python
from langchain_core.output_parsers import StrOutputParser

ouput_parser = StrOutputParser()
```

```python
chain = prompt | llm | ouput_parser
```

```python
result = chain.invoke({"question": "저는 식당에서 음식을 주문하고 싶어요."})
result
```

```python
print(result)
```

### JsonOutputParser

```python
from langchain_core.output_parsers import JsonOutputParser

parser = JsonOutputParser()
parser
```

```python
parser.get_format_instructions()
```

```python
template = """
당신은 영어를 아주 쉽게 가르치는 친절한 영어 선생님입니다.
주어진 상황에 맞는 영어 회화를 작성해 주세요.
양식은 [FORMAT]을 참고해 주세요.

# 상황:
{question}

# Format:
{format_instructions}
"""
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    template=template,
    input_variables=["question"],
    partial_variables={
        "format_instructions": parser.get_format_instructions()
    }
)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash"
)
```

```python
chain = prompt | llm | parser
```

```python
result = chain.invoke({"question": "저는 식당에서 음식을 주문하고 싶어요."})
```

```python
import pprint

pprint.pprint(result)
```

### 일관성 있는 형식으로 출력하기(키가 변경되지 않게)

```python
from pydantic import BaseModel, Field

class EnglishConversation(BaseModel):
    title: str = Field(..., description="상황에 대한 제목")
    dialogue: list = Field(..., description="상황에 맞는 영어 대화 리스트")
    explanation: str = Field(..., description="대화에 대한 쉬운 설명")
```

```python
paser = JsonOutputParser(pydantic_object=EnglishConversation)
paser
```

```python
template = """
당신은 영어를 아주 쉽게 가르치는 친절한 영어 선생님입니다.
주어진 상황에 맞는 영어 회화를 작성해 주세요.
양식은 [FORMAT]을 참고해 주세요.
모든 키 이름은 정확히 이 양식과 일치해야 합니다.

# 상황:
{question}

# Format:
{format_instructions}

# 예시:
{{
    "title": "저녁 식사 정하기",
    "dialogue": [
        {{"speaker": "A", "text": "What should we have for dinner?"}},
        {{"speaker": "B", "text": "I'm craving some pizza."}},
    ],
    "explanation": "친구와 저녁 메뉴를 정하는 상황입니다."
}}
"""
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    template=template,
    input_variables=["question"],
    partial_variables={
        "format_instructions": parser.get_format_instructions()
    }
)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3-flash-preview"
)
```

```python
chain = prompt | llm | parser
```

```python
result = chain.invoke({"question": "저는 식당에서 음식을 주문하고 싶어요."})
```

```python
import pprint

pprint.pprint(result)
```
