### Document loader

> https://docs.langchain.com/oss/python/integrations/document_loaders/pdfplumber

```python
# uv add langchain-community pdfplumber
from langchain_community.document_loaders import PDFPlumberLoader

loader = PDFPlumberLoader("../data/Employee_Benefits_Guide_2026_v1.pdf")
documents = loader.load()
```

```python
print(documents[3].page_content)
```

```python
print(documents[3].metadata)
```
