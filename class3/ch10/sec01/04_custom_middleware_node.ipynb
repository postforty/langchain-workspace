```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.agents.middleware import before_agent
from langchain.messages import AIMessage

@before_agent(can_jump_to = ["end"])
def validate_agent(state, runtime):
    human_message = state["messages"][-1]

    if "금지어" in human_message.content:
        print("금지어 감지")
        return {
            "messages": [AIMessage(content="현재 금지어가 포함되어 있습니다. 주의하세요.")],
            "jump_to": "end"
        }
    
    return None
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model="google_genai:gemini-2.5-flash",
    middleware=[validate_agent],
)
```

```python
agent.invoke(
    {"messages": [{"role": "user", "content": "이런 금지어 같은 경우를 봤나?"}]},
)
```
